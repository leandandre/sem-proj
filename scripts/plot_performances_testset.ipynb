```python
import numpy as np
from pathlib import Path
import json
from matplotlib import pyplot as plt
```

```python
PROJECT_ROOT = Path.cwd().parent
IMPORT_DIR = PROJECT_ROOT / "reports" / "metrics"
TARGET_DIR = PROJECT_ROOT / "reports" / "figure"
```

```python
### ctx free fully supervised L=20, S=5
dict_ctxfree_fullysuperv_l20s5 = IMPORT_DIR / "ctxfree_fullysupervised_results_stage2.json"
with open(dict_ctxfree_fullysuperv_l20s5, "r") as f:
    results_ctxfree_fullysuperv_l20s5 = json.load(f)
results_ctxfree_fullysuperv_l20s5_p1 = results_ctxfree_fullysuperv_l20s5["1.0"]
y = results_ctxfree_fullysuperv_l20s5_p1["stage2_per_class_f1"]
x = ["Wake", "N1", "N2", "N3", "REM"]
plt.figure(figsize=(10, 6))
plt.bar(x, y)
plt.xlabel("\nSleep Stage", fontsize=20)
plt.ylabel("F1 Score\n", fontsize=20)
plt.tick_params(axis='both', labelsize=16)
# plt.title("Per-Class F1 Scores for Context-Free Fully Supervised Model (L=20, S=5)")
plt.ylim(0, 1)
plt.axhline(y=results_ctxfree_fullysuperv_l20s5_p1["stage2_mf1"], color='r', linestyle='--', label='Macro F1 Score')
for i, f1 in enumerate(y):
    plt.text(i, f1 + 0.02, round(f1, 3), ha='center', fontsize=18)
plt.text(1, 0.72, f"{round(results_ctxfree_fullysuperv_l20s5_p1['stage2_mf1'], 3)}", ha='center', color='r', fontsize=18)
plt.legend(fontsize=12)
# plt.show()
plt.savefig(TARGET_DIR / "ctxfree_fullysuperv_l20s5_p1.0_perclass_f1.pdf", dpi=300, bbox_inches='tight')
```

```python
### ctx sensitive fully supervised L=20, S=5
dict_ctxsensitive_fullysuperv_l20s5 = IMPORT_DIR / "ctxsensitive_fullysupervised_results_stage2.json"
with open(dict_ctxsensitive_fullysuperv_l20s5, "r") as f:
    results_ctxsensitive_fullysuperv_l20s5 = json.load(f)
results_ctxsensitive_fullysuperv_l20s5_p1 = results_ctxsensitive_fullysuperv_l20s5["1.0"]
y = results_ctxsensitive_fullysuperv_l20s5_p1["stage2_per_class_f1"]
x = ["Wake", "N1", "N2", "N3", "REM"]
plt.figure(figsize=(10, 6))
plt.bar(x, y)
plt.xlabel("\nSleep Stage", fontsize=20)
plt.ylabel("F1 Score\n", fontsize=20)
plt.tick_params(axis='both', labelsize=16)
# plt.title("Per-Class F1 Scores for Context-Sensitive Fully Supervised Model (L=20, S=5)")
plt.ylim(0, 1)
plt.axhline(y=results_ctxsensitive_fullysuperv_l20s5_p1["stage2_mf1"], color='r', linestyle='--', label='Overall Macro F1 Score')
for i, f1 in enumerate(y):
    plt.text(i, f1 + 0.02, round(f1, 3), ha='center', fontsize=18)
plt.text(1, 0.78, f"{round(results_ctxsensitive_fullysuperv_l20s5_p1['stage2_mf1'], 3)}", ha='center', color='r', fontsize=18)
plt.legend(fontsize=12)
# plt.show()
plt.savefig(TARGET_DIR / "ctxsensitive_fullysuperv_l20s5_p1.0_perclass_f1.pdf", dpi=300, bbox_inches='tight')
```

```python
### ctx free fully finetuned L=20, S=5
dict_ctxfree_finetuning_l20s5 = IMPORT_DIR / "ctxfree_finetuning_results_stage2.json"
with open(dict_ctxfree_finetuning_l20s5, "r") as f:
    results_ctxfree_finetuning_l20s5 = json.load(f)
results_ctxfree_finetuning_l20s5_p1 = results_ctxfree_finetuning_l20s5["1.0"]
y = results_ctxfree_finetuning_l20s5_p1["stage2_per_class_f1"]
x = ["Wake", "N1", "N2", "N3", "REM"]
plt.figure(figsize=(10, 6))
plt.bar(x, y)
plt.xlabel("\nSleep Stage", fontsize=20)
plt.ylabel("F1 Score\n", fontsize=20)
plt.tick_params(axis='both', labelsize=16)
# plt.title("Per-Class F1 Scores for Context-Free Fully Finetuned Model (L=20, S=5)")
plt.ylim(0, 1)
plt.axhline(y=results_ctxfree_finetuning_l20s5_p1["stage2_mf1"], color='r', linestyle='--', label='Overall Macro F1 Score')
for i, f1 in enumerate(y):
    plt.text(i, f1 + 0.02, round(f1, 3), ha='center', fontsize=18)
plt.text(1, 0.70, f"{round(results_ctxfree_finetuning_l20s5_p1['stage2_mf1'], 3)}", ha='center', color='r', fontsize=18)
plt.legend(fontsize=12)
# plt.show()
plt.savefig(TARGET_DIR / "ctxfree_finetuning_l20s5_p1.0_perclass_f1.pdf", dpi=300, bbox_inches='tight')
```

```python
### ctx free finetuned vs fully supervised overall mf1
dict_ctxfree_finetuning_l20s5 = IMPORT_DIR / "ctxfree_finetuning_results_stage2.json"
with open(dict_ctxfree_finetuning_l20s5, "r") as f:
    results_ctxfree_finetuning_l20s5 = json.load(f)

dict_ctxfree_fullysuperv_l20s5 = IMPORT_DIR / "ctxfree_fullysupervised_results_stage2.json"
with open(dict_ctxfree_fullysuperv_l20s5, "r") as f:
    results_ctxfree_fullysuperv_l20s5 = json.load(f)

mf1_finetuned = []
mf1_fullysuperv = []

for key in results_ctxfree_finetuning_l20s5:
    nested_dict = results_ctxfree_finetuning_l20s5[key]
    mf1_score = nested_dict['stage2_mf1']
    mf1_finetuned.append(mf1_score)

for key in results_ctxfree_fullysuperv_l20s5:
    nested_dict = results_ctxfree_fullysuperv_l20s5[key]
    mf1_score = nested_dict['stage2_mf1']
    mf1_fullysuperv.append(mf1_score)

p = [0.01, 0.05, 0.1, 0.2, 0.5, 1.0]
p = [x * 100 for x in p]  # convert to percentage

plt.figure(figsize=(10, 6))
plt.plot(p, mf1_finetuned, marker='x', linewidth=2, markersize=8, label='SSL Pretrain + FT', color='green')
plt.plot(p, mf1_fullysuperv, marker='o', linewidth=2, markersize=8, label='Supervised from Scratch', color='red')
plt.xlabel('\nPercentage of labeled training data (%)', fontsize=18)
plt.ylabel('Macro F1 Score\n', fontsize=18)
plt.legend(fontsize=16)
plt.tick_params(axis='both', labelsize=16)
plt.tight_layout()
plt.grid(True)
# plt.show()
plt.savefig(TARGET_DIR / 'ctxfree_finetuning_vs_fullysupervised_l20s5_varying_p.pdf', dpi=300, bbox_inches='tight')
```
