```python
# run this cell in either case to import all necessary libraries
import matplotlib.pyplot as plt
from pathlib import Path
import json

# Data: Fraction of labeled data vs. Macro F1 scores for three methods (label ratio is not actively reserved when using the p fraction of labeled data)

# PROJECT_ROOT = Path(__file__).resolve().parents[1] #works only for python scripts
PROJECT_ROOT = Path.cwd().parent
print(PROJECT_ROOT)
TARGET_DIR = PROJECT_ROOT / "plots"

p = [0.01, 0.05, 0.1, 0.2, 0.5, 1.0]
```

```python
### run if you want the context-free model plot STAGE1 ###
### context-free model, these results are from the powerpoint SSL results.pptx (stage1) ###
p = [0.01, 0.05, 0.1, 0.2, 0.5, 1.0]
mf1_linprobe = [0.4850, 0.5659, 0.5652, 0.5789, 0.5953, 0.5977]
mf1_finetuned = [0.5219, 0.5711, 0.6261, 0.6395, 0.6368, 0.6443]
mf1_fullysuperv = [0.4974, 0.5405, 0.5567, 0.6091, 0.6339, 0.6547]

# absolute values plot
plt.figure(figsize=(10, 6))
plt.plot(p, mf1_linprobe, marker='+', linewidth=2, markersize=8, label='SSL Pretrain Linear Probing', color='blue')
plt.plot(p, mf1_finetuned, marker='x', linewidth=2, markersize=8, label='SSL Pretrain + FT', color='green')
plt.plot(p, mf1_fullysuperv, marker='o', linewidth=2, markersize=8, label='Supervised from Scratch', color='red')
plt.xlabel('\nPercentage of Labeled Data Used (%)\n', fontsize=18)
plt.ylabel('\nMacro F1 Score\n', fontsize=18)
# plt.title('Linear Probing vs. SSL Fine-Tuning vs. Fully-Supervised End-to-End\n', fontsize=18)
plt.legend(fontsize=16)
plt.tick_params(axis='both', labelsize=16)
plt.tight_layout()
plt.grid(True)
# plt.show()
plt.savefig(TARGET_DIR / 'lin-probe_vs_finetuned_vs_fullysupervised_mf1_ctx-free_stage1.pdf', dpi=300, bbox_inches='tight')

# relative improvement plot stage1
p = [0.01, 0.05, 0.1, 0.2, 0.5, 1.0]
p = [x * 100 for x in p]  # convert to percentage
rel_impr_finetuned =[]
for fine, fully in zip(mf1_finetuned, mf1_fullysuperv):
    rel_impr_finetuned.append((fine - fully) / fully * 100)
plt.figure(figsize=(10, 6))
plt.plot(p, rel_impr_finetuned, marker='x', linewidth=2, markersize=8, color='green')
plt.xlabel('\nPercentage of labeled training data used (%)\n', fontsize=18)
# plt.xlabel('\nFraction of Labeled Data\n', fontsize=18)
plt.ylabel('\nRelative Improvement MF1 Score (%)\n', fontsize=18)
plt.tick_params(axis='both', labelsize=16)
plt.tight_layout()
plt.grid(True)
# plt.show()
plt.savefig(TARGET_DIR / 'finetuned_vs_fullysupervised_rel_impr_mf1_ctx-free_stage1.pdf', dpi=300, bbox_inches='tight')
```

```python
### run if you want the context-free model plot STAGE2 ###
PERF_DIR = PROJECT_ROOT / "reports" / "metrics"
with open(PERF_DIR / "ctxfree_finetuning_results_stage2.json", 'r') as f:
    res_finetuning_dict = json.load(f)
# print(res_finetuning_dict)
mf1_finetuned = []
for key in res_finetuning_dict:
    nested_dict = res_finetuning_dict[key]
    mf1_score = nested_dict['stage2_mf1']
    mf1_finetuned.append(mf1_score)
print(mf1_finetuned)

with open(PERF_DIR / "ctxfree_fullysupervised_results_stage2.json", 'r') as f:
    res_fullysuperv_dict = json.load(f)
# print(res_fullysuperv_dict)
mf1_fullysuperv = []
for key in res_fullysuperv_dict:
    nested_dict = res_fullysuperv_dict[key]
    mf1_score = nested_dict['stage2_mf1']
    mf1_fullysuperv.append(mf1_score)
print(mf1_fullysuperv)

# let's plot the results now
p = [0.01, 0.05, 0.1, 0.2, 0.5, 1.0]
p = [x * 100 for x in p]  # convert to percentage
plt.figure(figsize=(10, 6))
plt.plot(p, mf1_finetuned, marker='x', linewidth=2, markersize=8, label='SSL Pretrain + FT', color='green')
plt.plot(p, mf1_fullysuperv, marker='o', linewidth=2, markersize=8, label='Supervised from Scratch', color='red')
plt.xlabel('\nPercentage of Labeled Data Used (%)\n', fontsize=18)
plt.ylabel('\nMacro F1 Score\n', fontsize=18)
# plt.title('Linear Probing vs. SSL Fine-Tuning vs. Fully-Supervised End-to-End\n', fontsize=18)
plt.legend(fontsize=16)
plt.tick_params(axis='both', labelsize=16)
plt.tight_layout()
plt.grid(True)
# plt.show()
plt.savefig(TARGET_DIR / 'finetuned_vs_fullysupervised_mf1_ctx-free_stage2.pdf', dpi=300, bbox_inches='tight')

# relative improvement plot stage2
p = [0.01, 0.05, 0.1, 0.2, 0.5, 1.0]
p = [x * 100 for x in p]  # convert to percentage
rel_impr_finetuned =[]
for fine, fully in zip(mf1_finetuned, mf1_fullysuperv):
    rel_impr_finetuned.append((fine - fully) / fully * 100)
plt.figure(figsize=(10, 6))
plt.plot(p, rel_impr_finetuned, marker='x', linewidth=2, markersize=8, color='green')
plt.xlabel('\nPercentage of labeled training data used (%)\n', fontsize=18)
plt.ylabel('\nRelative Improvement MF1 Score (%)\n', fontsize=18)
plt.tick_params(axis='both', labelsize=16)
plt.tight_layout()
plt.grid(True)
# plt.show()
plt.savefig(TARGET_DIR / 'finetuned_vs_fullysupervised_rel_impr_mf1_ctx-free_stage2.pdf', dpi=300, bbox_inches='tight')
```
